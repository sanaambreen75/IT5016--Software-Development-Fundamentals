```python
# Create a Book class
class Book:

    # This function sets up the book details
    def __init__(self, title, author):
        self.title = title              # Store the book title
        self.author = author            # Store the author's name
        self.available = True           # Book is available at the start


# Create a Member class
class Member:

    # This function sets up the member details
    def __init__(self, name):
        self.name = name                # Store the member's name
        self.borrowed_books = []        # Create an empty list for borrowed books


# Create a Library class
class Library:

    # This function sets up the library
    def __init__(self):
        self.books = []                 # List to store all books
        self.members = []               # List to store all members

    # Function to add a book to the library
    def add_book(self, book):
        self.books.append(book)         # Add the book to the books list
        print("Book added successfully.")

    # Function to add a member to the library
    def add_member(self, member):
        self.members.append(member)     # Add the member to the members list
        print("Member added successfully.")

    # Function to borrow a book
    def borrow_book(self, title, member):

        # Check each book in the library
        for book in self.books:

            # Check if the book title matches
            if book.title == title:

                # Check if the book is available
                if book.available:

                    # Change the book status to borrowed
                    book.available = False

                    # Add the book to the member's borrowed books
                    member.borrowed_books.append(book)

                    print(member.name, "borrowed", book.title)

                else:
                    # Tell the user that the book is already borrowed
                    print("Book is already borrowed.")

                return

        # If the book cannot be found
        print("Book not found.")

    # Function to return a book
    def return_book(self, title, member):

        # Check the books borrowed by the member
        for book in member.borrowed_books:

            # Check if the title matches
            if book.title == title:

                # Change the book status back to available
                book.available = True

                # Remove the book from the member's borrowed list
                member.borrowed_books.remove(book)

                print(member.name, "returned", book.title)

                return

        # If the member did not borrow the book
        print("This book was not borrowed by this member.")

    # Function to display all books
    def display_books(self):

        print("\nBooks in Library:")

        # Go through every book in the library
        for book in self.books:

            # Check the availability of the book
            if book.available:
                status = "Available"
            else:
                status = "Borrowed"

            # Display the book details and status
            print(book.title, "-", book.author, "-", status)


# Create a Library object
library = Library()

# Create Book objects
book1 = Book("Harry Potter", "J.K. Rowling")
book2 = Book("The Hobbit", "J.R.R. Tolkien")

# Create a Member object
member1 = Member("Sana")

# Add the books to the library
library.add_book(book1)
library.add_book(book2)

# Add the member to the library
library.add_member(member1)

# Display all books
library.display_books()

# Borrow Harry Potter
library.borrow_book("Harry Potter", member1)

# Display books after borrowing
library.display_books()

# Return Harry Potter
library.return_book("Harry Potter", member1)

# Display books after returning
library.display_books()
```
